```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.lines as mlines

# Define the lines and stations
lines = {
    "Jubliee line": ["Stanmore", "Baker street", "Waterloo", "Canada Waters", "Canary warf"],
    "Elizabeth line": ["Woolwich", "Canary warf", "Liverpool street", "Paddington", "Heathrow 2"],
    "DLR": ["Canary warf", "South Quay", "Mudchute", "Cutty sark", "Lewisham"],
    "Piccadilly": ["Acton town", "Baston Manor", "Hounslow west", "Hatton Cross", "Heathrow 2"],
    "Northern": ["Kennington", "Waterloo", "Tottenham court", "Google street", "Warren Street"],
}

# Define the distances between stations (in miles)
distances = {
    ("Stanmore", "Baker street"): 1,
    ("Baker street", "Waterloo"): 3.4,
    ("Waterloo", "Canada Waters"): 3.1,
    ("Canada Waters", "Canary warf"): 3,
    ("Woolwich", "Canary warf"): 5,
    ("Canary warf", "Liverpool street"): 4,
    ("Liverpool street", "Paddington"): 4,
    ("Paddington", "Heathrow 2"): 16,
    ("Canary warf", "South Quay"): 1.7,
    ("South Quay", "Mudchute"): 0.7,
    ("Mudchute", "Cutty sark"): 6.7,
    ("Cutty sark", "Lewisham"): 2.5,
    ("Acton town", "Baston Manor"): 19.5,
    ("Baston Manor", "Hounslow west"): 28.5,
    ("Hounslow west", "Hatton Cross"): 2,
    ("Hatton Cross", "Heathrow 2"): 2,
    ("Kennington", "Waterloo"): 1.3,
    ("Waterloo", "Tottenham court"): 10,
    ("Tottenham court", "Google street"): 8,
    ("Google street", "Warren Street"): 8.2,
}

# Create the graph
graph = nx.Graph()

# Add edges from lines to the graph, and include the distance attribute
for line, stations in lines.items():
    for i in range(len(stations) - 1):
        # Get the distance from the dictionary if it exists
        station_pair = (stations[i], stations[i + 1])
        reverse_pair = (stations[i + 1], stations[i])
        distance = distances.get(station_pair) or distances.get(reverse_pair)

        if distance:
            graph.add_edge(stations[i], stations[i + 1], line=line, distance=distance)

# Ensure the graph is connected
is_connected = nx.is_connected(graph)
print(f"\nIs the graph connected? {is_connected}")

# Create a color map for the lines
line_colors = {
    "Jubliee line": "#f69697",
    "Elizabeth line": "#FEF8DD",
    "DLR": "lightblue",
    "Piccadilly": "pink",
    "Northern": "lightgreen",
}

# Color the edges based on the lines
edge_colors = [line_colors[graph[u][v]['line']] for u, v in graph.edges()]

# Color the nodes based on the lines they belong to
node_colors = []
for node in graph.nodes():
    # Find which lines the node belongs to
    node_lines = [line for line, stations in lines.items() if node in stations]
    if len(node_lines) == 1:
        # Color for distinct stations
        node_colors.append(line_colors[node_lines[0]])
    else:
        # Neutral color for stations served by multiple lines
        node_colors.append("#bca89f")

for line in lines:
    print(f"\n{line}:")
    for u, v, data in graph.edges(data=True):
        if data['line'] == line:
            print(f"  Distance between {u} and {v}: {data['distance']} Miles")


# Draw the graph with colored nodes and edges
plt.figure(figsize=(12, 8))

# Set a fixed seed for layout consistency
pos = nx.spring_layout(graph, seed=42, k=0.5)  # Adjust k for better node spacing

# Draw the nodes (stations) with colors based on the lines
nx.draw(
    graph,
    pos,
    with_labels=True,
    node_color=node_colors,  # Color stations based on the lines they belong to
    node_size=2000,
    font_size=8,
    edge_color=edge_colors,  # Color edges according to their line
)

# Add labels with distances on the edges
edge_labels = nx.get_edge_attributes(graph, 'distance')
nx.draw_networkx_edge_labels(
    graph,
    pos,
    edge_labels=edge_labels,
    font_size=8,
    font_color='black',
)

# Create Line2D objects for the legend (lines)
legend_lines = [mlines.Line2D([0], [0], color=line_colors[line], lw=4) for line in line_colors]

# legend for multi-line stations (stations shared by multiple lines)
same_line_legend = mlines.Line2D([0], [0], color="#bca89f", lw=4, label="Same line station")

#Label for the distance
plt.text(0.5, -0.05, "All the distances are in miles", ha='center', va='center', transform=plt.gca().transAxes, fontsize=10)


# Add the legends to the plot
plt.legend(legend_lines + [same_line_legend], list(line_colors.keys()) + ["Same line station"], title="Lines", loc="upper right")

plt.title("Transport Network of London Underground Services with Distances")
plt.show()
```
